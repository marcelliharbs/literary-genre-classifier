```python
import pandas as pd

df = pd.read_csv(
    "gutenberg_books.csv",
    dtype={"book_id": str}
)

df["texto"] = (
    df["trecho_1_clean"].fillna("") + " " +
    df["trecho_2_clean"].fillna("") + " " +
    df["trecho_3_clean"].fillna("")
)

print(f"Quantidade de livros: {len(df)}")
print(df.columns.tolist())
```

```python
for i in range(3):
    print("=" * 80)
    print(f"BOOK ID: {df.loc[i, 'book_id']}")
    print()
    print(df.loc[i, "texto"])
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Garante que não existam valores nulos
df["texto"] = (
    df["texto"]
    .fillna("")
    .astype(str)
)

vectorizer = TfidfVectorizer()

X_tfidf = vectorizer.fit_transform(
    df["texto"]
)

print("Dimensão da matriz TF-IDF:", X_tfidf.shape)

# transformar os textos em uma matriz de pesos TF-IDF
# resultando em uma matriz com dimensões de 312 linhas (livros) por 17.336 colunas 
# (termos/palavras únicas).
```
