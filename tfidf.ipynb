```python
import pandas as pd
import ast
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity


df = pd.read_csv(
    "gutenberg_books.csv",
    dtype={"book_id": str}
)

df["tokens"] = df["tokens_clean"].apply(ast.literal_eval)

df["texto_tfidf"] = df["tokens"].apply(
    lambda tokens: " ".join(tokens)
)
```

```python
for i in range(3):
    print("=" * 80)
    print(f"BOOK ID: {df.loc[i, 'book_id']}")
    print()
    print(df.loc[i, "tokens_clean"])
```

```python
vectorizer = TfidfVectorizer()

X_tfidf = vectorizer.fit_transform(df["texto_tfidf"])

print("Dimensão da matriz TF-IDF:", X_tfidf.shape)

# transformar os textos em uma matriz de pesos TF-IDF
# resultando em uma matriz com dimensões de 312 linhas (livros) por 17.336 colunas
# (termos/palavras únicas).
```

```python
def buscar(consulta, quantidade=10):
    consulta_tfidf = vectorizer.transform([consulta])

    similaridades = cosine_similarity(
        consulta_tfidf,
        X_tfidf
    ).flatten()

    indices = similaridades.argsort()[::-1]

    resultados = df.iloc[indices[:quantidade]].copy()
    resultados["similaridade"] = similaridades[indices[:quantidade]]

    return resultados
```

```python
resultados = buscar(
    "war politics",
    quantidade=20
)

resultados[
    ["book_id", "similaridade"]
]
```
